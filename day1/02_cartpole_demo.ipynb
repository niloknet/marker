Gymnasium 라이브러리를 사용해 CartPole 환경을 처음 실행해봅니다.

## 목표
- Gymnasium 환경 생성 방법 이해
- CartPole-v1의 기본 구조 파악
- `reset()`과 `step()` API 사용법 익히기

## 1. Gymnasium 설치 및 환경 생성

```python
import gymnasium as gym

# CartPole 환경 생성: 막대를 세운 채로 유지하는 게임
env = gym.make("CartPole-v1")

print("환경 생성 완료:", env.spec.id)
```

## 2. 환경 초기화 (reset)

```python
# 환경을 초기 상태로 리셋
state, info = env.reset()

labels = ["카트 위치", "카트 속도", "막대 각도", "막대 각속도"]
print("초기 상태 (observation):")
for i, (val, label) in enumerate(zip(state, labels)):
    print(f"  [{i}] {label:10s}: {val:+.6f}")
print("info:", info)
```

## 3. 한 스텝 실행 (step)

```python
# 행동: 0=왼쪽으로 밀기, 1=오른쪽으로 밀기
action = 1  # 오른쪽으로 밀기

next_state, reward, terminated, truncated, info = env.step(action)

labels = ["카트 위치", "카트 속도", "막대 각도", "막대 각속도"]
print("다음 상태 (observation):")
for i, (val, label) in enumerate(zip(next_state, labels)):
    print(f"  [{i}] {label:10s}: {val:+.6f}")
print("보상:", reward)
print("에피소드 종료(terminated):", terminated)
print("시간 초과(truncated):", truncated)
```

## 4. 랜덤 에이전트로 한 에피소드 실행

```python
import time
import imageio

# GIF 저장을 위해 rgb_array 모드로 환경 생성 (프레임 수집용)
env_episode = gym.make("CartPole-v1", render_mode="rgb_array")
state, info = env_episode.reset()
total_reward = 0
step_count = 0
labels = ["카트 위치", "카트 속도", "막대 각도", "막대 각속도"]
frames = [env_episode.render()]  # 초기 상태 프레임

done = False
while not done:
    action = env_episode.action_space.sample()  # 랜덤 행동
    state, reward, terminated, truncated, info = env_episode.step(action)
    frames.append(env_episode.render())  # 매 스텝 프레임 수집
    done = terminated or truncated
    total_reward += reward
    step_count += 1

    # 스텝마다 출력
    action_str = "왼쪽" if action == 0 else "오른쪽"
    print(f"[스텝 {step_count}] 행동: {action_str} ({action})")
    for i, (val, label) in enumerate(zip(state, labels)):
        print(f"  [{i}] {label:10s}: {val:+.6f}")
    print(f"  보상: {reward}, 종료: {done}")
    print("-" * 40)

    time.sleep(0.4)  # 사람이 볼 수 있는 속도 (초)

print(f"에피소드 종료: {step_count} 스텝, 총 보상 {total_reward}")

# 끝난 에피소드를 GIF로 저장
imageio.mimsave("last_episode.gif", frames, fps=30, loop=0)
env_episode.close()
```

```python
from IPython.display import Image, display
print("방금 끝난 에피소드 재생 (GIF):")
display(Image("last_episode.gif"))
```

## 정리
1. `gym.make("CartPole-v1")`로 환경 생성
2. `env.reset()` → 초기 상태와 info 반환
3. `env.step(action)` → (next_state, reward, terminated, truncated, info)
4. CartPole은 매 스텝 +1 보상, 막대가 쓰러지면 종료
