# FrozenLake 환경 소개

FrozenLake-v1의 기본 사용법을 익힙니다.

## 상태, 행동, 보상 개요

FrozenLake-v1(4x4)에서:
- **상태**: 0~15번 칸 (총 16개). 0=시작(S), 15=목표(G), 그 외 4x4 격자 위치.
- **행동**: 0=왼쪽, 1=아래, 2=오른쪽, 3=위 (이동이 맵 밖이면 제자리).
- **보상**: 매 스텝 0, 목표(G) 도달 시 1, 구멍(H)에 빠지면 0(에피소드 종료).

```python
import gymnasium as gym
env = gym.make("FrozenLake-v1", map_name="4x4")

# 상태: 0~nS-1 (4x4 → 16개)
print("【상태】")
print(f"  개수: {env.observation_space.n} (0 ~ {env.observation_space.n - 1})")
print(f"  의미: 칸 번호 (0=시작, 15=목표, 나머지=얼음/구멍 위치)")

# 행동: 0=LEFT, 1=DOWN, 2=RIGHT, 3=UP
print("\n【행동】")
print(f"  개수: {env.action_space.n}")
for i, name in enumerate(["LEFT", "DOWN", "RIGHT", "UP"]):
    print(f"  {i}: {name}")

# 보상: step마다 0, 목표 도달 1, 구멍 0
print("\n【보상】")
print("  매 스텝: 0")
print("  목표(G) 도달: 1")
print("  구멍(H) 낙사: 0 (에피소드 종료)")
env.close()
```

```python
import gymnasium as gym

# 4x4 맵: S=시작, F=얼음, H=구멍, G=목표
env = gym.make("FrozenLake-v1", map_name="4x4")
state, info = env.reset()
print("초기 상태(칸 번호):", state)
print("4x4 맵: 0=S, 15=G, H=구멍")
```

```python
# 맵 시각화
desc = env.unwrapped.desc
for row in desc:
    print(" ".join(c.decode() if hasattr(c, 'decode') else str(c) for c in row))
```

## Pygame 렌더 (한 장면)

FrozenLake 환경을 Pygame으로 렌더링하여 한 프레임을 표시합니다.

```python
import gymnasium as gym
import matplotlib.pyplot as plt

# render_mode="rgb_array" → Pygame 기반 렌더, numpy 배열로 반환
env = gym.make("FrozenLake-v1", map_name="4x4", render_mode="rgb_array")
state, _ = env.reset()
frame = env.render()

plt.figure(figsize=(5, 5))
plt.imshow(frame)
plt.axis("off")
plt.title("FrozenLake 4x4 (시작 상태)")
plt.tight_layout()
plt.show()
env.close()
```

```python
# 랜덤 에이전트로 한 에피소드
state, _ = env.reset()
done = False
steps = 0
while not done:
    action = env.action_space.sample()
    state, reward, terminated, truncated, _ = env.step(action)
    done = terminated or truncated
    steps += 1
print(f"에피소드 종료: {steps} 스텝, 보상={reward}")
env.close()
```
