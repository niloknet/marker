# FrozenLake env.P (전이 확률) 분석

env.unwrapped.P를 통해 전이 확률 구조를 분석합니다.

```python
import gymnasium as gym

env = gym.make("FrozenLake-v1", map_name="4x4")
P = env.unwrapped.P
```

```python
# P[s][a] = [(prob, next_state, reward, terminated), ...]
# 상태 0, 행동 1(아래)의 전이
s, a = 0, 1
print(f"P[{s}][{a}] (상태{s}에서 행동{a}=아래):")
for prob, next_s, reward, term in P[s][a]:
    print(f"  확률 {prob:.2f} -> 다음상태 {next_s}, 보상 {reward}, 종료 {term}")
```

```python
# FrozenLake: slippery=True(기본)면 1/3 확률로 의도와 다른 방향으로 이동
# 모든 (s,a)에 대해 전이 출력
print("전이 요약 (상태0 기준):")
for a in range(4):
    actions = ["왼쪽", "아래", "오른쪽", "위"]
    print(f"  행동 {a}({actions[a]}): {P[0][a]}")
```

## 정리
P[s][a]는 (확률, 다음상태, 보상, 종료여부) 튜플의 리스트입니다.
